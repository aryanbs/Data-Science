# Task 1.1 Environmental Sensor Analysis

A sensor records hourly temperatures over 72 hours. Some readings are corrupted and contain unrealistic values.

```python
import numpy as np

np.random.seed(42)

raw_temperatures = np.random.normal(loc=24.0, scale=3.0, size=72)

# Simulated corrupt readings
raw_temperatures[[5, 12, 28, 45, 60]] = [99.9, -15.0, 88.8, -50.0, 102.4]
```

#### Tasks

1. Create a boolean mask for valid temperatures.

```python
bool_mask = (raw_temperatures >= 15.0) & (raw_temperatures <= 40.0)
```

2. Replace all invalid values with the median of the valid temperatures.

```python
# Extracting valid temperatures
valid_temperatures = raw_temperatures[bool_mask]
```

```python
print(valid_temperatures)
```

```python
# Computing the median of valid values
median_value = np.median(valid_temperatures)
```

```python
print(median_value)
```

```python
# Replacing invalid values using vectorization
cleaned_temperatures = np.where(bool_mask, raw_temperatures, median_value)
```

```python
print(cleaned_temperatures)
```

3. Calculate and print Mean, SD, Variance

```python
mean_val = np.round(np.mean(cleaned_temperatures), 2)
std_val = np.round(np.std(cleaned_temperatures), 2)
var_val = np.round(np.var(cleaned_temperatures), 2)
```

```python
print("Mean:", mean_val)
print("Standard Deviation:", std_val)
print("Variance:", var_val)
```

# Part 2: Sliding Median Filter

```python
signal = np.array([10, 11, 99, 12, 13, -45, 14, 15, 16])
```

```python
#Creating a copy
```

```python
filtered_signal = signal.copy()
```

2. Write a loop that iterates from index 1 to 7.
3. For each position:

```python
for i in range(1, len(signal) - 1):
    
    # Extract window [i-1, i, i+1]
    window = signal[i-1:i+2]
    
    # Compute median
    median_value = np.median(window)
    
    # Update filtered signal
    filtered_signal[i] = median_value
```

Print both:

```python
print("Original Signal:", signal)
print("Filtered Signal:", filtered_signal)
```

The median filter removes spikes in the signal by replacing each value with the median of its neighboring values within a sliding window. In this case, extreme values such as 99 and -45 are significantly higher or lower than their surrounding values. When the median is computed over a window of three elements, these outliers do not affect the result because the median selects the middle value after sorting, effectively ignoring extreme deviations.

This makes the median filter particularly effective for removing impulse noise or sudden spikes. In contrast, using the mean would be influenced by these extreme values, causing the filtered result to still be distorted. Therefore, the median filter provides a more robust and reliable smoothing technique for noisy signals with outliers.
